# HuggingFace EARS Data Integration Demo

This notebook demonstrates how to easily access and visualize underwater acoustic data from the GoMRI-17 dataset hosted on HuggingFace, using the `dolphain` library.

## Prerequisites
You need to have `huggingface_hub` installed.

```python
import dolphain
from dolphain.plotting import plot_overview
import matplotlib.pyplot as plt

# Configure matplotlib for notebook
%matplotlib inline
```

## 1. Load a Random File
The easiest way to get started is to load a random file from the dataset. This requires just one line of code.

```python
# Fetch a random file from HuggingFace
ears = dolphain.EARS()

# Display basic info
print(ears)
ears.info()
```

## 2. Visualize the Data
We can use `dolphain.plotting.plot_overview` to see the waveform and spectrogram of the loaded file.

```python
# Plot waveform and spectrogram
# Note: We pass the full metadata dictionary to plot_overview
plot_overview(ears.metadata)
plt.show()
```

## 3. Load a Specific File
You can also request a specific file by its filename.

```python
# Load a specific file known to have interesting features
ears_specific = dolphain.EARS(filename="7178E5DC.200")
print(ears_specific)

# Visualize it
plot_overview(ears_specific.metadata)
plt.show()
```
